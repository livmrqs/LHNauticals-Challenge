# Questão 2 — Geração automática do schema

## Objetivo

A partir dos arquivos CSV fornecidos pelo ERP, gerar automaticamente um
schema compatível com PostgreSQL.

A solução foi implementada utilizando exclusivamente bibliotecas da
biblioteca padrão do Python 3, conforme solicitado.

## Estratégia

O processo consiste em:

1. identificar automaticamente os arquivos CSV do diretório de entrada;
2. extrair as colunas a partir dos cabeçalhos;
3. analisar os valores não nulos para inferir seus tipos;
4. mapear os tipos encontrados para tipos compatíveis com PostgreSQL;
5. gerar uma instrução `CREATE TABLE` para cada CSV;
6. consolidar todas as instruções em um único arquivo `schema.sql`.

Nenhuma biblioteca externa é utilizada durante o processo.

```python
from pathlib import Path
import subprocess
import sys


PROJECT_ROOT = Path.cwd().parent

SCRIPT_PATH = PROJECT_ROOT / "src" / "data" / "generate_schema.py"
INPUT_DIR = PROJECT_ROOT / "data" / "raw"
OUTPUT_PATH = PROJECT_ROOT / "sql" / "schema.sql"
```

```python
result = subprocess.run(
    [
        sys.executable,
        str(SCRIPT_PATH),
        "--input-dir",
        str(INPUT_DIR),
        "--output",
        str(OUTPUT_PATH),
    ],
    check=True,
    capture_output=True,
    text=True,
)

print(result.stdout)
```

```python
#Validação
csv_files = list(INPUT_DIR.glob("*.csv"))

schema_content = OUTPUT_PATH.read_text(encoding="utf-8")

created_tables = schema_content.count("CREATE TABLE IF NOT EXISTS")

print(f"Arquivos CSV encontrados: {len(csv_files)}")
print(f"Tabelas geradas: {created_tables}")

assert created_tables == len(csv_files)
```

### Resultado da validação

Foram identificados 24 arquivos CSV no diretório de origem e 24 tabelas foram
geradas no arquivo `schema.sql`.

A validação confirma que todos os arquivos disponíveis foram contemplados pelo
processo de geração automática do schema.
